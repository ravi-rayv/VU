## 2. Your own study area

Upload a KML/KMZ/GeoJSON/zipped shapefile, or give a path. You need your own Earth Engine project (see `docs/earth_engine.md`).

```python
import ee
EE_PROJECT = 'your-ee-project-id'
ee.Authenticate()            # once per machine / Colab session
ee.Initialize(project=EE_PROJECT)
```

```python
AOI_FILE = 'examples/demo_area.kml'   # <- your file
# In Colab you can upload instead:
# from google.colab import files; up = files.upload(); AOI_FILE = list(up)[0]

from vertical25d import read_aoi, Inputs, area_km2
aoi = read_aoi(AOI_FILE)
print(f'{area_km2(aoi):.2f} km²')
aoi.explore() if hasattr(aoi, 'explore') else aoi.plot()
```

```python
s = Settings(grid_m=500, n_sim=200, run_change=False)
res = run(Inputs(aoi=aoi, ee_project=EE_PROJECT, name='my_area'), s, out_dir='outputs')
print('Results:', res.zip_path)
```

```python
res.grid.explore(column='VSI_excess', cmap='RdBu', tiles='CartoDB positron') if res.grid is not None else None
```
